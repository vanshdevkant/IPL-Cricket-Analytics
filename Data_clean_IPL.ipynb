## Dataset Note
This project uses IPL data (2008-2026) from Kaggle. The 2026 season data may be partial/incomplete as it was likely captured before the season's final matches were played, so 2026 statistics (e.g., Orange Cap standings) may not match the actual final season results.

```python
# Data Cleaning 
# step 1
print(matches.isnull().sum())  #it will show in columns  missing/empty values how much 
```

```python
# clean missing/empty in columns

matches = matches.fillna({
    "city":"Unknown",
    "match_number":"Unknown",
    "win_by_runs":0,
    "win_by_wickets":0,
    "player_of_match":" No Result"
    })
```

```python
# now not missing/empty check bellow
print(matches.isnull().sum())
```

```python
# checking Duplicates values 
print(matches.duplicated().sum())
```

```python
# but don't have dupicates values
# suppose u have duplicates values use bellow fuction
matches = matches.drop_duplicates()
# tahn use shape funtion to check reomve or not
print(matches.shape)

```

```python
print(matches.dtypes)
```

```python
# convert match_date column into date bcz now not in date
matches["match_date"] = pd.to_datetime(matches["match_date"])
print(matches.dtypes)
```

```python
# we can be see which types data stored in column 
print(matches['season'].unique())
```

```python
# use replace functions 
matches["season"] = matches["season"].replace("2020/21","2020")
```

```python
# convert season column into int

matches["season"] = matches["season"].astype(int)
```

```python
print(matches["season"].unique())
```

```python
print(matches['city'].unique())
```

```python
# text columns ka extra space automatically remove karta hai 

text_columns = matches.select_dtypes(include='object').columns
for col in text_columns:
    matches[col] = matches[col].str.strip()
```

```python
print(matches['city'].unique())
```

```python
# for Save Data clean

matches.to_csv("matches_clean.csv", index = False)
print("saved")
```

```python
# clean automatically sapce

text_columns = deliveries.select_dtypes(include='object').columns
for col in text_columns:
    deliveries[col] = deliveries[col].str.strip()
```
